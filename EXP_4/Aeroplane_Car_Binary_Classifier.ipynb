**Reading Images into R**

```r
pics <- converted_files

my_pic <- list()

for (i in 1:12) {
  my_pic[[i]] <- readImage(pics[i])
}
```

```r
print(my_pic[[1]])

display(my_pic[[1]])

display(my_pic[[8]])

summary(my_pic[[1]])

hist(my_pic[[12]])

hist(my_pic[[2]])

str(my_pic)
```

**Image Preprocessing**

```r
for (i in 1:12) {
  my_pic[[i]] <- resize(my_pic[[i]], 28, 28)
}

str(my_pic)
```

```r
for (i in 1:12) {
  my_pic[[i]] <- array_reshape(
    my_pic[[i]],
    c(28 * 28 * 3)
  )
}

str(my_pic)
```

**Partitioning Data into Train & Test Sets**

```r
trainX <- NULL

for (i in 1:5) {
  trainX <- rbind(trainX, my_pic[[i]])
}

for (i in 7:11) {
  trainX <- rbind(trainX, my_pic[[i]])
}

str(trainX)

testX <- rbind(
  my_pic[[6]],
  my_pic[[12]]
)

trainY <- c(
  rep(0, 5),
  rep(1, 5)
)

testY <- c(0, 1)

trainLabels <- to_categorical(trainY)

testLabels <- to_categorical(testY)
```

**Building the Sequential Neural Network**

```r
model <- keras_model_sequential() %>%
  layer_dense(
    units = 256,
    activation = "relu",
    input_shape = c(2352)
  ) %>%
  layer_dense(
    units = 128,
    activation = "relu"
  ) %>%
  layer_dense(
    units = 2,
    activation = "softmax"
  )

summary(model)
```

**Compiling the Model**

```r
model %>% compile(
  loss = "binary_crossentropy",
  optimizer = optimizer_rmsprop(),
  metrics = c("accuracy")
)
```

**Training the Model**

```r
history <- model %>% fit(
  trainX,
  trainLabels,
  epochs = 30,
  batch_size = 32,
  validation_split = 0.2
)
```

```r
plot(history)
```

**Model Evaluation**

```r
evaluation <- model %>% evaluate(
  trainX,
  trainLabels
)

print(evaluation)
```

**Training Predictions**

```r
pred_prob <- predict(model, trainX)

pred <- apply(pred_prob, 1, which.max) - 1

print(pred)
```

**Training Confusion Matrix**

```r
table(
  Predicted = pred,
  Actual = trainY
)
```

**Prediction Probabilities**

```r
cbind(
  pred_prob,
  Predicted = pred,
  Actual = trainY
)
```

**Testing Predictions on Unseen Data**

```r
pred_test_prob <- predict(model, testX)

pred_test <- apply(pred_test_prob, 1, which.max) - 1

print(pred_test)
```

**Testing confusion matrix**

```r
table(
  Predicted = pred_test,
  Actual = testY
)
```
