```r
numbers <- c(2, 4, 6, 7)

for (i in numbers) {
  print(i^2)
}
```

```r
x <- c(2, 4, 6, 8, 10, 12)

count <- 0

for (xval in x) {
  if (xval / 2 > 3) {
    count <- count + 1
  }
}

print(count)
```

```r
child <- c("child1", "child2", "child3")
sweet <- c("sweet1", "sweet2", "sweet3")

for (x in child) {
  for (y in sweet) {
    print(paste(x, y))
  }
}
```

```r
drink <- c("coffee", "lemonade", "tea", "juice")

for (x in drink) {

  if (x == "tea") {
    break
  }

  print(x)
}
```

```r
drink <- c("coffee", "lemonade", "tea", "juice")

for (x in drink) {

  if (x == "lemonade") {
    next
  }

  print(x)
}
```

```r
i <- 1

while (i < 10) {
  print(i^2)
  i <- i + 2
}
```

```r
sum <- 0
number <- 22

while (number <= 25) {
  sum <- sum + number
  number <- number + 1
}

print(sum)
```

```r
i <- 1

repeat {

  print(i^2)

  i <- i + 2

  if (i > 10) {
    break
  }
}
```

```r
abc <- function(x) {
  x^2
}

abc(3)
abc(6)
abc(9)
```

```r
abc <- function(x, y) {
  x^2 + y^2
}

abc(3, 4)
abc(10, 10)
abc(-2, -3)
```

```r
abc <- function() {

  for (i in 1:3) {
    print(i^3)
  }
}

abc()
```
